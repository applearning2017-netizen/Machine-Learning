```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder ,MinMaxScaler,StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB, MultinomialNB, BernoulliNB
from sklearn.linear_model import LogisticRegression,LogisticRegressionCV
from sklearn.model_selection import train_test_split, cross_val_score,GridSearchCV
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report,f1_score,recall_score,precision_score
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/applearning2017-netizen/data/refs/heads/main/insurance.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
le = LabelEncoder()
df['sex'] = le.fit_transform(df['sex'])
df['smoker'] = le.fit_transform(df['smoker'])
df['region'] = le.fit_transform(df['region'])
#df = pd.get_dummies(df, columns=['Sex', 'Embarked','Pclass'])
print(len(df))
df.head()
```

```python
X=df.drop(columns=['smoker'])
X
```

```python
y=df['smoker']
#scaler = StandardScaler()
#X = scaler.fit_transform(X)
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 10)
```

```python
logr = LogisticRegression()
logr.fit(X_train,y_train)
predicted = logr.predict(X_test)
```

```python
cm = confusion_matrix(y_test, predicted)

print('Confusion matrix\n\n', cm)
```

```python
print(classification_report(y_test, predicted))
```

```python
print("Accuracy:", accuracy_score(y_test, predicted))
```

```python
logr.get_params()
```

```python
#Value of C can't be negtive so
param_grid = {
    'C': np.linspace(0.01, 8, 10),
    'penalty': ['l1', 'l2'],
    'solver': ['lbfgs', 'liblinear', 'newton-cg', 'newton-cholesky', 'sag', 'saga'],
    'max_iter': [50,100, 200, 300,400,500]
}


log_reg = LogisticRegression()
grid_lr = GridSearchCV(log_reg, param_grid, cv=3, scoring='accuracy', )
grid_lr.fit(X_train, y_train)
```

```python
# Print best parameters
print("Best parameters:", grid_lr.best_params_)
y_pred_lr = grid_lr.best_estimator_.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred_lr))
```
